```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv('/content/netflix_titles.csv')

df.head()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x='type')

plt.title('Distribution of Movies and TV Shows')
plt.xlabel('Content Type')
plt.ylabel('Count')

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(data=df, x='release_year', bins=30, kde=True)

plt.title('Distribution of Netflix Release Years')
plt.xlabel('Release Year')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.countplot(data=df, x='rating', hue='type')

plt.title('Netflix Content Type vs Rating')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(data=df, x='type', y='release_year')

plt.title('Release Year vs Content Type')
plt.xlabel('Content Type')
plt.ylabel('Release Year')

plt.show()
```
